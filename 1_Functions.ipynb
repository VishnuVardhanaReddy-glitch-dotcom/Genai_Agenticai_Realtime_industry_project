```python
def add_sub_multi_div(a,b):
    c = a+b
    d = a-b
    e = a*b
    f = a/b
    return c,d,e,f
```

```python
add_sub_multi_div(15,10)
```

```python
def arg(name, age):
    print("name is -->",name, "age is -->",age)
```

```python
arg("ram", 12)
```

```python
def kwarg(name, age = 10):
    print(name, age)
```
